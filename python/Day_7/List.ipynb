```python
#  creating list by using [] -- square brackets
List_1 = ["amit", 12, 13.5, True, False, 3+2j]
```

```python
print(type(List_1))
print(List_1)
```

```python
#  creating list using list() method
list_2 = list((1, 2, 3, 4, 5.9, "Amit", True, False, None, 5 + 6j))
```

```python
print(type(list_2))
print(list_2)
```

```python
# Indexing -- ye kis position pe kon sa fruit h.
fruits = ["Apple", "banana", "cherry", "kiwi", "Pineapple", "Strawberry" ]


# Left to right --- +ve indexing -- counting starts from zero(0).

fruits[2]
```

```python
# Right to left --> -ve indexing -- counting starts with -1
fruits[-2]
```

```python
# slicing == getting a range of items == [Start: Stop: Step]
# Start -- this will be included in your output.
# Stop -- this will be excluded in your output.
# Step -- this will be escaped.

fruits[::2]
```

```python
fruits
```

```python
fruits[::-1]
```

```python
## List methods

# Number of items inside the list -- len()
len(fruits)
```

```python
# Modifying the list items

fruits[1] = "Watermelon"
print(fruits)
```
