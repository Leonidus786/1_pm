# Booleans -- True and False

```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool(()))
print(bool([]))
print(bool({}))
```

```python
print(bool(True))
print(bool("NDLS"))
print(bool(5))
print(bool("Amit"))
print(bool((1,)))
print(bool([1,2,3]))
print(bool({"A": 5}))
```
