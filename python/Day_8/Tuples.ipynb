##### Introduction to Tuples
Explanation:

Tuples are ordered collections of items that are immutable.
They are similar to lists, but their immutability makes them different.


```python
## creating a tuple
empty_tuple= ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
my_tuple = (1,)

my_list = [1]

print(type(my_tuple))
print(type(my_list))
print(my_tuple)
print(my_list)
```

```python
lst=list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
numbers=tuple([1,2,3,4,5,6])
numbers
```

```python
list((1,2,3,4,5,6))
```

```python
mixed_tuple=(1,"Hello World",3.14, True, 2+3j)
print(mixed_tuple)
```

```python
## Accessing Tuple Elements -- Indexing

numbers
```

```python
print(numbers[2])  # +ve indexing -- Left to right
print(numbers[-1]) # -ve indexing -- Right to left
```

```python
# slicing
numbers[0:4]
```

```python
# reverse
numbers[::-1]
```

```python
print(numbers)
print(mixed_tuple)
```

```python
## Tuple Operations

concatenation_tuple=numbers + mixed_tuple
print(concatenation_tuple)
```

```python
print(mixed_tuple * 3)
```

```python
numbers *3
```

```python
## Immutable Nature Of Tuples
## Tuples are immutable, meaning their elements cannot be changed once assigned.

lst=[1,2,3,4,5]
print(lst)

lst[1]="Abhi"
print(lst)
```

```python
numbers[1]="Abhi"
```

```python
numbers
```

```python
## Tuple Methods
print(numbers.count(1))
print(numbers.index(3))
```

```python
## Packing and Unpacking tuple
## packing
packed_tuple = 1,"Hello",3.14,5
print(packed_tuple)
```

```python
a,b,c,d = 1, "hello python", 3, 4+2j


a
```

```python
##unpacking a tuple
a,b,*c=packed_tuple

print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## Nested Tuple
## Nested List
lst=[[1,2,3,4],[6,7,8,9],[1,"Hello",3.14,"c"]]
lst[0][0:3]
```

```python
lst=[[1,2,3,4],[6,7,8,9],(1,"Hello",3.14,"c")]
# lst[2][0:3]

lst[1][-1]
```

```python
nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))

## access the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
print(nested_tuple)
len(nested_tuple)
```

```python
numbers=(1,2,3,4,5,6)
numbers
```

```python
for i in numbers:
    print(i*3)
```

```python
## iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
